# Card Balance & Scoring Framework

## 📊 **Essential Stats for FaB Card Evaluation**

### **Power Level Metrics**
1. **Resource Efficiency** - Cost vs Impact ratio
2. **Tempo Value** - Immediate board impact 
3. **Card Advantage** - Net card draw/loss
4. **Threat Level** - How fast it pressures opponent
5. **Versatility** - Multiple use cases
6. **Synergy Rating** - How well it works with archetype

### **Key Balance Factors**

#### **Equipment Balance**
- **Cost vs DEF ratio** (sweet spot: 2-4 Teklo Energy per 1 DEF)
- **Transform value** (upgrade should feel meaningful)
- **Battleworn sustainability** (can it survive multiple combats?)

#### **Item Balance** 
- **Activation cost** (1-3 Teklo Energy typical)
- **Impact per activation** (immediate vs long-term)
- **Resource generation** (counters, energy, cards)

#### **Attack Balance**
- **Damage per resource** (6-8 attack per 2 Teklo Energy)
- **Go Again accessibility** (conditional vs guaranteed)
- **Defensive counterplay** (can opponent reasonably block?)

---

## 🎯 **Scoring System (1-10 Scale)**

### **Individual Card Scores**
- **Power Level**: 1 (weak) → 10 (format-defining)
- **Complexity**: 1 (simple) → 10 (very complex)
- **Fun Factor**: 1 (boring) → 10 (exciting interactions)
- **Balance**: 1 (broken) → 10 (perfectly balanced)

### **Archetype Coherence**
- **Theme Consistency**: How well it fits Teklovossen's identity
- **Mechanical Synergy**: Works with other cards in set
- **Power Curve**: Appropriate for competitive level

## ⚖️ **Balance Guidelines for Teklovossen**

### **Teklo Energy Economy**
- **Generation**: 1-2 per turn typical
- **Spending**: Most abilities cost 1-2 energy
- **Storage**: Limited, encourage active play

### **Nanite Counter Balance**
- **Generation Rate**: 1-2 per turn max
- **Spending Value**: 1 counter = minor effect
- **Cap Consideration**: Prevent infinite accumulation

### **AI Item Synergies**
- **Activation Rewards**: Bonus for using multiple AIs
- **Resource Cost**: Higher individual costs, synergy discounts
- **Diminishing Returns**: Prevent AI spam being optimal

### **Evo Power Curve**
- **Base → Evo**: Should feel like meaningful upgrade
- **Multi-Evo**: Exponential power, exponential cost
- **Transform Cost**: 3-5 Teklo Energy typical for major upgrade

````python
# Card Balance Calculator for Teklovossen
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Define card evaluation metrics
class CardEvaluator:
    def __init__(self):
        self.cards = []
        
    def add_card(self, name, card_type, cost, power_level, complexity, fun_factor, balance, theme_fit, notes=""):
        card = {
            'Name': name,
            'Type': card_type,
            'Cost': cost,
            'Power_Level': power_level,
            'Complexity': complexity,
            'Fun_Factor': fun_factor,
            'Balance': balance,
            'Theme_Fit': theme_fit,
            'Overall_Score': np.mean([power_level, fun_factor, balance, theme_fit]),
            'Notes': notes
        }
        self.cards.append(card)
        
    def get_dataframe(self):
        return pd.DataFrame(self.cards)
    
    def analyze_balance(self):
        df = self.get_dataframe()
        if len(df) == 0:
            print("No cards to analyze yet!")
            return
            
        print("=== CARD BALANCE ANALYSIS ===")
        print(f"Average Power Level: {df['Power_Level'].mean():.1f}/10")
        print(f"Average Balance Score: {df['Balance'].mean():.1f}/10") 
        print(f"Average Overall Score: {df['Overall_Score'].mean():.1f}/10")
        
        # Flag problematic cards
        high_power_low_balance = df[(df['Power_Level'] >= 8) & (df['Balance'] <= 5)]
        if len(high_power_low_balance) > 0:
            print("\n⚠️  HIGH POWER, LOW BALANCE CARDS:")
            for idx, card in high_power_low_balance.iterrows():
                print(f"- {card['Name']}: Power {card['Power_Level']}, Balance {card['Balance']}")
                
        return df
    
    def plot_balance_scatter(self):
        df = self.get_dataframe()
        if len(df) == 0:
            return
            
        plt.figure(figsize=(10, 6))
        scatter = plt.scatter(df['Power_Level'], df['Balance'], 
                            c=df['Fun_Factor'], cmap='viridis', 
                            s=df['Complexity']*20, alpha=0.7)
        
        plt.xlabel('Power Level')
        plt.ylabel('Balance Score')
        plt.title('Teklovossen Card Balance Analysis')
        plt.colorbar(scatter, label='Fun Factor')
        
        # Add card names as labels
        for idx, card in df.iterrows():
            plt.annotate(card['Name'], (card['Power_Level'], card['Balance']), 
                        xytext=(5, 5), textcoords='offset points', fontsize=8)
        
        # Add balance zones
        plt.axhline(y=5, color='red', linestyle='--', alpha=0.5, label='Balance Threshold')
        plt.axvline(x=8, color='orange', linestyle='--', alpha=0.5, label='High Power Threshold')
        
        plt.legend()
        plt.grid(True, alpha=0.3)
        plt.show()

# Initialize evaluator
evaluator = CardEvaluator()

# Add your existing cards for analysis
print("Card Balance System initialized!")
print("Use evaluator.add_card() to score your designs")
print("Use evaluator.analyze_balance() to see results")
````

## ⚖️ **Balance Methodology: How to Fix Cards**

### **Step-by-Step Balance Process**

#### **1. Identify the Problem** 🔍
- **Too Cheap**: Effect too powerful for cost
- **No Limitations**: Unlimited uses or scaling
- **No Counterplay**: Opponent can't respond
- **Power Creep**: Strictly better than existing cards

#### **2. Apply the Right Fix** 🔧

| **Problem Type** | **Solution** | **Example** |
|------------------|--------------|-------------|
| **Free/Too Cheap** | Add or increase cost | Predator Logic: 0 → 1 Teklo Energy |
| **Unlimited Scaling** | Add caps or "once per turn" | Singularity Cannon: Max +4 attack |
| **No Resource Pressure** | Add decay or maintenance costs | Nanite Reactor: Lose counters at 5+ |
| **Auto-Include Power** | Add timing restrictions | "Once per turn when..." |
| **No Risk/Reward** | Add sacrifice/banish costs | Predator Logic: Banish AI Item |

#### **3. Preserve Identity** 🎯
- **Keep the core mechanic** that makes the card unique
- **Maintain synergy** with the archetype
- **Preserve flavor** and thematic elements

#### **4. Test New Balance Score** 📊
- **Power Level**: Should drop 1-2 points
- **Balance Score**: Should increase to 7+
- **Fun Factor**: Should stay same or improve
- **Overall**: Target 7.5+ average

```

Moderate Effect (+2 attack, card draw): 1-2 Teklo Energy

Game-Changing Effect: 3+ Teklo Energy
```

### **Attack Power Guidelines**
```
Base Attack Value: 4-6
Per 1 Teklo Energy: +2 Attack
Per Resource Spent: +1-2 Attack
Legendary Weapons: 6-8 base attack
```

### **Counter/Resource Guidelines**
```
Generation Rate: 1-2 per turn max
Spending Value: 1 counter = 1 Teklo Energy effect
Maximum Storage: 5-7 counters before decay
```
